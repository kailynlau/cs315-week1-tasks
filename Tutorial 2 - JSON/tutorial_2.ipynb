## Introducing JSON

Note: JSON stands for JavaScript Object Notation.

A key-value pair is separated by a colon, with a string key on the left in double quotes.

| JSON Data Type | Description |
|---|---|
| object | A collection of key-value pairs inside curly braces ({}) |
| array | A list of values wrapped in square brackets ([]) |
| string | Text wrapped in double quotes ("") |
| number | Integers or floating-point numbers |
| boolean | Either true or false without quotes |
| null | Represents a null value, written as null |

JSON *does not* support comments or trailing commas.

## Writing JSON With Python

Serialization: converting data into a JSON
Deserialization: the reverse

| Python | JSON | Allowed as JSON Key |
|---|---|---|
| dict | object | No |
| list | array | No |
| tuple | array | No |
| str | string | Yes |
| int | number | Yes |
| float | number | Yes |
| True/False | true/false | Yes |
| None | null | Yes |

```python
food_ratings = {"organic dog food": 2, "human food": 10}
dog_data = json.dumps(food_ratings) # Returns a JSON-formatted string
with open("hello_frieda.json", mode="w", encoding="utf-8") as write_file:
    json.dump(dog_data, write_file) # actually dumps to JSON file
```

```python
available_nums = {(1, 2): True, 3: False}
json.dumps(available_nums, skipkeys=True) # to skip any problematic keys and therefore avoid TypeError
# can also use sort_keys = True as well as some other params
```

## Interacting With JSON

It is possible to add a indent parameter to json.dumps for prettier formatting.

Useful to check if a JSON file is valid:
$ python -m json.tool dog_friend.json (run in terminal)

Can also add indents in terminal view: $ python -m json.tool hello_frieda.json --indent 2

(This is helpful to reduce / minimize the size of the JSON file, useful for large data files, and then adding back when necessary during viewing.)
